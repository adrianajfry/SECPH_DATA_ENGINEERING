```python
## Phase 1: Data Preparation - Person A (Matching & Cleaning)

# Tasks:
# 1. Merge `FuelConsumption.csv` (2014 emissions) and `Cars2025.csv` (2025 registrations)
# 2. Clean the merged dataset
```

```python
# Step 1: Import Libraries

import pandas as pd
import numpy as np
```

```python
# Step 2: Load Datasets

fuel_data = pd.read_csv('FuelConsumption.csv')  # 2014 emissions data
cars_data = pd.read_csv('cars_2025.csv')        # 2025 registration data

# Display first few rows to inspect
print("Fuel Consumption Data (2014):")
display(fuel_data.head(2))

print("\nCar Registrations Data (2025):")
display(cars_data.head(2))
```

```python
# Step 3: Dataset Matching

# Standardize column names for merging
fuel_data['maker'] = fuel_data['MAKE'].str.upper().str.strip()
fuel_data['model'] = fuel_data['MODEL'].str.upper().str.strip()
cars_data['maker'] = cars_data['maker'].str.upper().str.strip()
cars_data['model'] = cars_data['model'].str.upper().str.strip()

# Now merge on the new standardized columns
merged_data = pd.merge(
    cars_data,
    fuel_data,
    on=['maker', 'model'],
    how='left',
    suffixes=('_2025', '_2014')
)

print("\nMerged Data Shape:", merged_data.shape)
merged_data.head(3)
```

```python
# Step 4: Initial Cleaning 

# Handle missing values - proper inplace operations
print("\nMissing Values Before Cleaning:")
display(merged_data.isnull().sum())

# Create a copy for cleaning to avoid chained assignment warnings
clean_data = merged_data.copy()

# Fill missing values without chained assignment
if 'ENGINESIZE' in clean_data.columns:
    clean_data['ENGINESIZE'] = clean_data['ENGINESIZE'].fillna(
        clean_data['ENGINESIZE'].median()
    )
    
if 'CYLINDERS' in clean_data.columns:
    clean_data['CYLINDERS'] = clean_data['CYLINDERS'].fillna(
        clean_data['CYLINDERS'].mode()[0]
    )

# Add emission data flag
clean_data['has_emission_data'] = ~clean_data['CO2EMISSIONS'].isna()

# Convert fuel consumption if column exists
if 'FUELCONSUMPTION_COMB' in clean_data.columns:
    clean_data['fuel_consumption_L100km'] = 235.21 / clean_data['FUELCONSUMPTION_COMB']

# Verify cleaning
print("\nMissing Values After Cleaning:")
display(clean_data.isnull().sum())

# Update merged_data reference
merged_data = clean_data
```

```python
# Step 5: Export Cleaned Data
merged_data.to_csv('Cleaned_Merged_Cars_FIXED.csv', index=False)
```

```python
# Final Verification
print("\nCleaned Data Sample:")
display(merged_data[['maker', 'model', 'CO2EMISSIONS', 'has_emission_data']].sample(5))
```
