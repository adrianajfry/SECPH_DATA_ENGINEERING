```python
def calculate_bmi(weight, height): # define new function and its parameters
    bmi = weight / (height * 2) # declare the formulae given
    if bmi < 18.5:
        category = "underweight"
    elif 18.5 <= bmi < 24.9:
        category = "normal"
    elif 25 <= bmi < 29.9:
        category = "overweight"
    else:
        category = "obese"
```

```python
def extract_year(matric_number): # define what we want to do (function)
    return matric_number[3:7] # only return index from 3 until before 7
```

```python
def check_faculty(matric_number):
    return matric_number.startwith("MAT") # only return if it's from math faculty
```

```python
def generate_matric_number(fac, year, student_id):
    return f"{fac}{year}{student_id:04d}" # 04d is a 4-digit unique number 
```

```python
import numpy as np

random_array = np.random.randint(1, 101, size=30)
print(random_array)
```

```python
reshaped_array = random_array.reshape(5, 6)
print(reshaped_array)
```

```python
column_index = 2
column = reshaped_array[:, column_index]
print(column)

column_index = 3
column_sum = reshaped_array[:, column_index].sum() # sum is a method, don't put it behind array name
print(column_sum)

new_column = reshaped_array[:, 2] + reshaped_array[:, 3]
reshaped_array_with_sum = np.column_stack((reshaped_array, new_column))
print(reshaped_array_with_sum)
```
